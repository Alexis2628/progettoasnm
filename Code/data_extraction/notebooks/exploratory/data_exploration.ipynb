# Esplorazione iniziale dei dati

In questo notebook analizziamo i dataset raccolti dalle query **AI**, **ML** e **ChatGPT** su Threads. Tutti i file `data.csv` sono presenti nella cartella `Code/data_extraction/data/raw`.


```python
import pandas as pd
from pathlib import Path
```

```python
base_path = Path('..') / 'data' / 'raw'
paths = {name: base_path / name / 'data.csv' for name in ['AI','ML','ChatGPT']}
dfs = {name: pd.read_csv(path) for name, path in paths.items()}
for name, df in dfs.items():
    print(name, df.shape)
```

```python
for name, df in dfs.items():
    display(df.head())
```

```python
# Numero di righe, colonne e colonne disponibili
import pandas as pd
summary = pd.DataFrame({name: {'rows': len(df), 'columns': len(df.columns), 'fields': df.columns.tolist()} for name, df in dfs.items()}).T
summary
```

```python
# Lunghezza media delle caption e like medi
import numpy as np
stats = []
for name, df in dfs.items():
    caption_len = df['caption'].astype(str).str.len()
    stats.append({'dataset': name, 'mean_caption_len': caption_len.mean(), 'mean_like_count': df['like_count'].mean()})
pd.DataFrame(stats)
```
