```python
from pathlib import Path
import pandas as pd
DATA_DIR = Path.cwd().parent.parent / 'data' / 'processed' 
PATH_1 = DATA_DIR / 'post_data' / 'total_post1.csv'
PATH_2 = DATA_DIR / 'post_data' / 'total_post2.csv'
PATH_3 = DATA_DIR / 'post_data' / 'total_post3.csv'
```

```python
df1 = pd.read_csv(PATH_1)
df2 = pd.read_csv(PATH_2)
df3 = pd.read_csv(PATH_3)
df = pd.concat([df1, df2, df3], ignore_index=True)
```

```python
n = df.shape[0] //4
```

```python
df1 = df[:n]
df2 = df[n:2*n]
df3 = df[2*n:3*n]
df4 = df[3*n:]
df1.to_csv(DATA_DIR / 'total_post1.csv', index=False)
df2.to_csv(DATA_DIR / 'total_post2.csv', index=False)
df3.to_csv(DATA_DIR / 'total_post3.csv', index=False)
df4.to_csv(DATA_DIR / 'total_post4.csv', index=False)
```
